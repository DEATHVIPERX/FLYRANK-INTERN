## 1. My rule and its reason codes

*Write the rule in plain words first. Then the reason codes it can output.*

**Rule in plain words:** A content item is worth reviewing if it has been online for a long time, has a high number of impressions, and its average position has decreased recently, indicating it might be losing visibility.

**Reason Codes:**
- `HIGH_IMPRESSIONS_OLD_CONTENT`: Content with high impressions and older publication date.
- `POSITION_DECLINE`: Content whose average position has worsened significantly.
- `HIGH_IMPRESSIONS_OLD_CONTENT_AND_POSITION_DECLINE`: Content exhibiting both conditions.
- `LOW_SCORE`: Content that doesn't meet the criteria for review based on this rule.

```python
import pandas as pd
import numpy as np

# Load the starter dataset
df = pd.read_csv('flyrank-ml-internship-starter/data/raw/content_refresh_anonymized.csv')

print('Dataset loaded. Displaying first 5 rows and info:')
display(df.head())
df.info()
```

```python
# 2. Code it as a transparent score.

# Define thresholds and conditions for the rule:
# 'long time online': Using 'days_since_last_update'. Let's consider content older than 180 days as 'long time online'.
# 'high number of impressions': Using 'impressions_90d'. Let's consider > 500 impressions as 'high'.
# 'average position has decreased recently': Using 'trend_direction' (as gsc_avg_position_prev30/90 are not in starter dataset).
# If trend_direction is 'down', it indicates position has decreased.

df['is_old_content'] = (df['days_since_last_update'] >= 180).astype(int)
df['has_high_impressions'] = (df['impressions_90d'] >= 500).astype(int)
df['position_declined'] = (df['trend_direction'] == 'down').astype(int)

# Combine conditions for scoring. We'll give a higher score to items that meet more criteria.
# For a transparent score, let's multiply. This means only items meeting ALL conditions get a score > 0 initially.
# And then multiply by impressions_90d to prioritize highly-trafficked declining content.
df['score'] = (df['is_old_content'] * df['has_high_impressions'] * df['position_declined'] * df['impressions_90d']).fillna(0)

print('Scores calculated. Displaying content with highest scores:')
display(df.sort_values(by='score', ascending=False).head())
```

```python
# 3. Attach reason codes.

# Initialize reason codes as empty strings or 'LOW_SCORE'
df['reason_code'] = 'LOW_SCORE'

# Apply reason codes based on conditions
df.loc[(df['is_old_content'] == 1) & (df['has_high_impressions'] == 1) & (df['position_declined'] == 0), 'reason_code'] = 'HIGH_IMPRESSIONS_OLD_CONTENT'
df.loc[(df['is_old_content'] == 0) & (df['has_high_impressions'] == 0) & (df['position_declined'] == 1), 'reason_code'] = 'POSITION_DECLINE'
df.loc[(df['is_old_content'] == 1) & (df['has_high_impressions'] == 0) & (df['position_declined'] == 1), 'reason_code'] = 'POSITION_DECLINE' # if old and declined, prioritize decline
df.loc[(df['is_old_content'] == 0) & (df['has_high_impressions'] == 1) & (df['position_declined'] == 1), 'reason_code'] = 'POSITION_DECLINE' # if high imp and declined, prioritize decline
df.loc[(df['is_old_content'] == 1) & (df['has_high_impressions'] == 1) & (df['position_declined'] == 1), 'reason_code'] = 'HIGH_IMPRESSIONS_OLD_CONTENT_AND_POSITION_DECLINE'

print('Reason codes attached. Displaying content with highest scores and their reason codes:')
display(df[df['score'] > 0].sort_values(by='score', ascending=False).head())
```

```python
# 4. Rank and evaluate at K.

# First, define the precision@k function
def precision_at_k(scores, labels, k):
    order = np.argsort(-np.asarray(scores))
    # Ensure k does not exceed the number of available labels
    k = min(k, len(labels))
    return np.asarray(labels)[order[:k]].mean()

# For this assignment (ML-07), we need to identify items for review.
# The 'is_declining_label' is not directly present in the starter dataset but can be derived from 'trend_direction'.
# Note: The skill mentions 'is_declining_label' is derived from 'trend_direction' and 'trend_pct'
# and should not be used as a feature, but it's appropriate for evaluating the baseline.

# Create 'is_declining_label' from 'trend_direction'
df['is_declining_label'] = (df['trend_direction'] == 'down')

# Evaluate precision@K
K = 20 # As per 'Top-20 Review'
precision_k = precision_at_k(df['score'], df['is_declining_label'], K)
base_rate = df['is_declining_label'].mean()

print(f"Precision@{K}: {precision_k:.2f}")
print(f"Base rate (proportion of declining content in dataset): {base_rate:.2f}")

# Prepare the ranked queue for output
ranked_queue = df.sort_values(by='score', ascending=False).reset_index(drop=True)

# Save the ranked queue to a CSV
output_dir = 'work/outputs'
os.makedirs(output_dir, exist_ok=True)
output_path = os.path.join(output_dir, 'baseline_action_score.csv')

ranked_queue.to_csv(output_path, index=False)
print(f"Ranked queue saved to {output_path}")

print('\nDisplaying the top 20 ranked items:')
display(ranked_queue.head(K))
```
